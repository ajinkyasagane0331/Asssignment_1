Q1:
A PivotTable is used to summarize, analyze, and reorganize large datasets quickly.

Q2:
To show total sales, go to Insert → PivotTable → New Worksheet, then drag Region to Rows and Sales to Values.

Q3:
For Conditional Formatting, go to Home, select the Total Sales column, then click Conditional Formatting → Greater Than, and enter the required value.

Q4:
Sorting rearranges the data and is used to rank data in ascending or descending order.
Filtering shows only specific values and hides the rest of the data.

Q5:
Create a PivotTable, drag Date to Rows and Sales to Values.
Then insert a chart, which automatically sets the X-axis and Y-axis.

Q6:
Go to the PivotTable, click PivotTable Analyze → Insert Slicer, select Product, then click OK.

Q7:
Dashboards are used for quick insights and a summarized view of data.

Q8:
Create an Excel dashboard by inserting a PivotTable to show Sum of Sales by Region and add a column chart for comparison.
Then insert a Product slicer and another PivotTable grouped by Month with a line chart to show the monthly sales trend, all on one worksheet.
